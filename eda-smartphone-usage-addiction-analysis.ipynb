
# **Load & Understand Data**



```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/kaggle/input/datasets/harilav/eda-on-smartphone-usage-dataset/Smartphone_Usage_And_Addiction_Analysis_7500_Rows.csv")
df.head()
```

# **Data Types & Stucture**

```python
df.info()
```

# **Statistical Summary**

```python
df.describe()
```

# **Missing Values**

```python
df.isnull().sum()
```

```python
df['addiction_level'].fillna(df['addiction_level'].mode()[0], inplace=True)
df.head()
```

```python
df.isnull().sum()
```

# **Duplicates**

```python
df.duplicated().sum()
df.drop_duplicates(inplace=True)
```

# **Univariate Analysis(ONE VARIABLE)**

```python
#Numerical
df['daily_screen_time_hours'].hist(edgecolor='black')
plt.grid(axis='x')
plt.title('Distribution of Daily Screen Time')
plt.xlabel('Daily Screen Time (hours)')
plt.ylabel('Frequency')
plt.show()
```

```python
#Categorical
df['gender'].value_counts()
df['gender'].value_counts().plot(kind='bar', edgecolor = 'black')
plt.title('Gender Distribution')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.grid(axis='y')
plt.show()
```

# **Outlier Detection**

```python
sns.boxplot(x=df['daily_screen_time_hours'])
plt.title('Boxplot of Daily Screen Time')
plt.xlabel('Daily Screen Time (hours)')
plt.show()
```

# **Bivariate Analisis(TWO VARIABLE)**

```python
plt.scatter(df['daily_screen_time_hours'], df['addiction_level'])
plt.title('Daily Screen Time vs Addiction Level')
plt.xlabel('Daily Screen Time (hours)')
plt.ylabel('Addiction Level')
plt.show()
```

```python
plt.scatter(df['age'], df['addiction_level'])
plt.title('Age vs Addiction Level')
plt.xlabel('Age')
plt.ylabel('Addiction Level')
plt.show()
```

# **Group Comparison**

```python
sns.boxplot(x='addiction_level', y='daily_screen_time_hours', data=df)
plt.title('Daily Screen Time by Addiction Level')
plt.show()
```

```python
avg_screen_time_by_stress = df.groupby('stress_level')['daily_screen_time_hours'].mean().reset_index()
sns.barplot(x='stress_level', y='daily_screen_time_hours', data=avg_screen_time_by_stress)
plt.title('Average Daily Screen Time by Stress Level')
plt.xlabel('Stress Level')
plt.ylabel('Average Daily Screen Time (hours)')
plt.show()
```

# **Correlation (RELATIONSHIP STRENGTH)**

```python
corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True)
plt.title('Correlation Heatmap')
plt.show()
```

# **Multivariate Analysis**

```python
sns.pairplot(df)
plt.show()
```
