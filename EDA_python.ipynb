# **Data Cleaning:**

```python
missing_vals = df.isnull().sum()
print(missing_vals[missing_vals > 0])
```

```python
df = df.dropna()
df
```

```python
for col in df.select_dtypes(include='str').columns:
    df[col] = df[col].astype('category')
```

```python
print("\nData Info after cleaning:")
print(df.info())
```

# Observation:
- Missing data handled; data types optimized.
- Ready for outlier detection and analysis.

# **Outlier Detection**

```python
for col in cols:
    plt.figure(figsize=(8, 4))
    sns.boxplot(x=df[col])
    plt.title(f'Boxplot of {col}')
    plt.show()
```

```python
num_cols = df.select_dtypes(include=['number']).columns
```

```python
for col in num_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    print(f'Number of outliers in {col}: {len(outliers)}')
```

# Obsevation
- For each numerical column, it calculates the outlier bounds.
- It filters and displays all rows where the value is outside these bounds.
- Shows the total count and the actual outlier observations.
